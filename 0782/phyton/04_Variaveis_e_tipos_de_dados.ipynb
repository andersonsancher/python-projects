## 2. O que é uma variável
Uma variável é um **nome** que damos a um valor para o podermos identificar e usar ao longo do programa. Sem variáveis, teríamos de repetir os mesmos valores literais em todo o lado.

````python
print("Bom dia, Maria")
print("Maria, qual e o teu plano para hoje?")
print("Ate logo, Maria")
````

Com uma variável, escreves uma vez e usas várias; para mudar o nome basta alterar uma linha:

````python
nome = "Maria"
print("Bom dia,", nome)
print(nome, "qual e o teu plano para hoje?")
print("Ate logo,", nome)
````

### 2.1. A variável como rótulo, não como caixa
Em C/C++/Java diz-se que uma variável é uma "caixa". Em Python é mais correto pensar numa variável como um **rótulo** (etiqueta) que se cola a um valor. É possível ter dois nomes a apontar para o mesmo valor:

````python
preco = 12.50
valor = preco   # agora 'valor' aponta para o mesmo numero que 'preco'
print(preco, valor)   # 12.5 12.5
````

> **Sem stress com este detalhe.** Por agora podes pensar nas variáveis como caixas e tudo correrá bem. A distinção só se torna importante mais tarde, com listas e estruturas mutáveis.

### 2.2. Atribuição: o operador `=`
Para criar uma variável usa-se o operador de atribuição `=`. À esquerda o nome; à direita o valor.

````python
idade = 25
nome = "Joao"
altura = 1.78
ativo = True
print(idade, nome, altura, ativo)
````

> **`=` não é igualdade matemática.** É uma ordem: "atribui ao nome da esquerda o valor da direita". Por isso `idade = 25` funciona, mas `25 = idade` dá erro. A comparação "é igual a?" faz-se com `==` (dois sinais de igual).

Em Python, ao contrário de C ou Java, **não declaras o tipo** da variável: escreves `idade = 25`, não `int idade = 25;`. O Python descobre o tipo sozinho.

### 2.3. Reatribuir e atribuir em cadeia
Uma variável pode receber um novo valor a qualquer momento. Cada atribuição substitui o valor anterior.

````python
x = 10
print(x)   # 10
x = 20
print(x)   # 20
x = x + 5
print(x)   # 25
````

A linha `x = x + 5` lê-se da direita para a esquerda: calcula `x + 5` e atribui o resultado a `x`.

````python
# Atribuir o mesmo valor a varias variaveis:
a = b = c = 0
print(a, b, c)   # 0 0 0

# Atribuir varios valores em paralelo:
x, y, z = 1, 2, 3
print(x, y, z)   # 1 2 3

# Trocar valores numa unica linha (tuple unpacking):
a = 1
b = 2
a, b = b, a
print(a, b)   # 2 1
````

## 3. Regras de nomenclatura
### 3.1. Regras obrigatórias (impostas pela linguagem)
- Só pode conter letras, dígitos e o sublinhado `_`.
- Não pode começar por um dígito.
- Distingue maiúsculas de minúsculas (`idade`, `Idade`, `IDADE` são diferentes).
- Não pode ser uma palavra reservada.

Exemplos **válidos**:

````python
nome = "Ana"
idade = 30
_temp = 0
valor_total = 99.99
x1 = 1
y2 = 2
print(nome, idade, _temp, valor_total, x1, y2)
````

Exemplos **inválidos** (dão `SyntaxError`):
```python
1ano = 2026          # comeca por digito
nome-cliente = "Ana" # contem hifen
valor total = 100    # contem espaco
if = 5               # 'if' e palavra reservada
```

### 3.2. Convenções recomendadas (PEP 8)
- **`snake_case`** para variáveis e funções.
- **`MAIUSCULAS`** para constantes (valores que não vão mudar).
- Nomes **claros e significativos**.

````python
# snake_case (bom)
numero_de_alunos = 25
preco_unitario = 12.50

# MAIUSCULAS para constantes
TAXA_IVA = 0.23
PI = 3.14159265

# Nomes claros vs pouco claros
quantidade = 4
total = preco_unitario * quantidade
print(total, TAXA_IVA, PI, numero_de_alunos)
````

> **Constantes em Python, só por convenção.** Ao contrário de C (`const`) ou Java (`final`), o Python não impede tecnicamente a alteração de uma constante. As maiúsculas são apenas um sinal: "este valor não está pensado para ser alterado".

## 4. Tipos de dados primitivos
Em Python todos os valores têm um **tipo**. Os tipos primitivos são apenas cinco.

### 4.1. Inteiros (`int`): números inteiros, sem parte decimal
Em Python 3 os inteiros não têm tamanho máximo.

````python
idade = 25
ano = 2026
temperatura = -3
print(idade, ano, temperatura)

# Inteiros gigantes sem problema:
print(2 ** 100)
````

**Bases numéricas** e **sublinhados** como separadores de legibilidade:

````python
decimal = 255          # base 10
binario = 0b11111111   # base 2  (prefixo 0b)
octal   = 0o377        # base 8  (prefixo 0o)
hexa    = 0xFF         # base 16 (prefixo 0x)
print(decimal, binario, octal, hexa)   # 255 255 255 255

populacao = 10_290_000   # sublinhados ignorados pelo Python
print(populacao)
````

### 4.2. Vírgula flutuante (`float`): números com parte decimal
O separador decimal é o **ponto**, não a vírgula.

````python
altura = 1.78
pi = 3.14159
velocidade_luz = 3e8       # notacao cientifica: 3 x 10^8
print(altura, pi, velocidade_luz)

# Pequenos erros de precisao (sistema IEEE 754, nao e bug do Python):
print(0.1 + 0.2)   # 0.30000000000000004
````

### 4.3. Booleanos (`bool`): apenas `True` ou `False`
Nota a maiúscula inicial (`true`/`false` em minúsculas não são reconhecidos).

````python
ativo = True
concluido = False
print(ativo, concluido)

# Curiosidade: bool e subclasse de int (True==1, False==0)
print(True + True)   # 2
````

### 4.4. Cadeias de carateres (`str`): texto
Delimitadas por aspas simples `'` ou duplas `"` (equivalentes). Aspas triplas para texto multilinha.

````python
nome = "Maria"
cidade = 'Santarem'
vazia = ""
frase1 = "O Joao disse 'bom dia'"
frase2 = 'Ela respondeu "ola"'
print(nome, cidade, frase1, frase2)

poema = """
Tudo o que escrevo,
tudo o que digo.
"""
print(poema)
````

### 4.5. O valor `None`
Representa a **ausência de valor**: não é "zero", não é uma string vazia, é simplesmente nada.

````python
resultado = None
print(resultado)   # None
````

> **`None` vs zero vs string vazia.** `0` é um inteiro (tem valor), `0.0` é um float, `""` é uma string vazia (tem valor), mas `None` é mesmo "nada".

### 4.6. Quadro-resumo dos tipos

| Tipo | Significado | Exemplos |
|---|---|---|
| `int` | Números inteiros | `0`, `1`, `-5`, `2**100` |
| `float` | Números com parte decimal | `1.5`, `-3.14`, `2e10` |
| `bool` | Verdadeiro ou falso | `True`, `False` |
| `str` | Texto | `"Ana"`, `'Olá'`, `""` |
| `NoneType` | Ausência de valor | `None` |

## 5. Verificar o tipo de uma variável
### 5.1. Função `type()`: devolve o tipo

````python
idade = 25
print(type(idade))   # <class 'int'>
print(type(3.14))    # <class 'float'>
print(type("Ola"))   # <class 'str'>
print(type(True))    # <class 'bool'>
print(type(None))    # <class 'NoneType'>
````

### 5.2. Função `isinstance()`: responde sim/não (`True`/`False`)

````python
idade = 25
print(isinstance(idade, int))   # True
print(isinstance(idade, str))   # False
print(isinstance("Maria", str)) # True

# isinstance considera heranca: bool e subclasse de int
print(isinstance(True, int))    # True
print(type(True) == int)        # False
````

## 6. Conversão entre tipos (*casting*)
Para cada tipo primitivo existe uma função de conversão: `int()`, `float()`, `str()`, `bool()`.
### 6.1. Conversões mais comuns

````python
# String -> numero
print(int("42"))      # 42
print(float("3.14"))  # 3.14

# Numero -> string
print(str(42))        # '42'

# Entre int e float (int() TRUNCA, nao arredonda)
print(float(10))      # 10.0
print(int(3.99))      # 3  (nao 4!)
print(round(3.99))    # 4  (round arredonda)
````

**Para booleano:** valores "vazios" ou "zero" são `False`; tudo o resto é `True`.

````python
print(bool(0))        # False
print(bool(1))        # True
print(bool(-5))       # True
print(bool(0.0))      # False
print(bool(""))       # False
print(bool("texto"))  # True
print(bool(None))     # False
````

### 6.2. Casos especiais e armadilhas
- `int("3.14")` falha com **`ValueError`**; converte primeiro para `float`: `int(float("3.14"))`.
- `int("vinte")` falha, por ser texto não numérico.
- Espaços antes/depois são perdoados: `int("  42  ")` → `42`.
- **Vírgula em vez de ponto** falha: `float("3,14")` dá `ValueError`. Substitui antes de converter.

````python
# Estrategia para a virgula decimal portuguesa:
print(int(float("3.14")))            # 3
print(int("  42  "))                 # 42
print(float("3,14".replace(",", ".")))  # 3.14
````

> **Porque os erros começam por `ValueError`.** Quando uma operação falha, o Python gera uma **exceção**. `ValueError` → o valor não é adequado; `TypeError` → os tipos não são compatíveis. O tratamento de exceções (`try/except`) fica para a UFCD 0785.

## 7. Tipagem dinâmica
Em Python o tipo está associado ao **valor**, não à variável. Por isso a mesma variável pode mudar de tipo ao longo do programa:

````python
valor = 10
print(valor, type(valor))   # 10 <class 'int'>

valor = "agora sou texto"
print(valor, type(valor))   # ... <class 'str'>

valor = 3.14
print(valor, type(valor))   # 3.14 <class 'float'>
````

> **Dinâmica ≠ fraca.** Python é dinâmico mas **forte**: aceita que `valor` seja primeiro um inteiro e depois uma string, mas **não** tolera `1 + "texto"`, isso dá erro. O JavaScript, esse, é dinâmico e fraco (`1 + "texto"` dá `"1texto"`).

## 8. Anotações de tipo (*type hints*): introdução
A sintaxe: depois do nome, dois pontos e o tipo.

````python
nome: str = "Maria"
idade: int = 30
altura: float = 1.65
ativo: bool = True
print(nome, idade, altura, ativo)
````

> **MUITO IMPORTANTE: *type hints* não fazem nada em tempo de execução.** São apenas indicativas. O Python aceita-as e ignora-as: `idade: int = 30` seguido de `idade = "trinta"` **não dá erro**. Servem para as IDEs e verificadores estáticos como `mypy`. Aplicam-se sobretudo a funções (UFCD 0784).

## 9. Resumo do tópico
- Uma variável é um **nome (rótulo)** associado a um valor; usa-se `=` para atribuir.
- Em Python não se declara o tipo: é descoberto a partir do valor.
- Tipos primitivos: `int`, `float`, `bool`, `str`, `NoneType` (`None`).
- `type()` devolve o tipo; `isinstance()` responde sim/não.
- Conversão com `int()`, `float()`, `str()`, `bool()`. `int()` **trunca**.
- Conversões inválidas geram `ValueError`.
- Python tem **tipagem dinâmica** mas é **fortemente** tipado.
- *Type hints* documentam, mas não têm efeito em tempo de execução.
