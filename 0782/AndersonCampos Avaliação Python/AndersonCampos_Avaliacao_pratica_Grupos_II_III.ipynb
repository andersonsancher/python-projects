## Questão 5 - Cartão de identificação  *(1,2 valores)*

Crie um programa que peça ao utilizador o seu **nome** e a sua **idade** e mostre a mensagem:

```
Ola <nome>, tens <idade> anos e nasceste por volta de <ano>.
```

O **ano de nascimento** deve ser calculado a partir do ano atual (**2026**) e da idade.

*Sugestão: guarde o ano atual numa variável.*

## Questão 6 - Retângulo  *(1,8 valores)*

Peça ao utilizador a **largura** e a **altura** (em cm) de um retângulo e calcule e mostre:

a) a **área** (largura × altura);
b) o **perímetro** (2 × (largura + altura)).

*As medidas podem ter casas decimais.*

````python
largura = float(input("Me de uma largura: " ))
altura = float(input("Me de uma altura: " ))
area = altura * largura

print(f"A área de um retângulo: {area} ")

perimetro = 2 * (largura + altura)

print(f"O perímetro do retângulo é: {perimetro} ")

````

## Questão 7 - Média de notas  *(1,8 valores)*

Peça **três notas** (de 0 a 20) de um formando e calcule:

a) a **soma** das três notas;
b) a **média** das três notas, com casas decimais.

## Questão 8 - Conversor de tempo  *(2,2 valores)*

Peça um valor em **segundos** (número inteiro) e converta-o para **horas, minutos e segundos**, usando os operadores `//` (divisão inteira) e `%` (resto).

Mostre, por exemplo:

```
7325 segundos = 2 h 2 m 5 s
```

````python
total_segundos = int(input("me de alguns segundos para converter para minutos e horas: "))

horas = total_segundos // 3600
resto = total_segundos % 3600
minutos = resto // 60
segundos = resto % 60

print(f"{total_segundos} segundos = {horas} h {minutos} m {resto} s")
````

## Questão 9 - Temperatura  *(1,8 valores)*

Peça uma temperatura em **graus Celsius** e converta-a para **Fahrenheit**, sabendo que:

```
F = C * 9 / 5 + 32
```

Mostre o resultado numa frase clara, por exemplo:

```
25.0 °C correspondem a 77.0 °F
```

````python
Celsius = float(input("Me diga quantos Graus Celsius estão: "))
Fahrenheit = Celsius * 9 / 5 + 32

print(f"Convertendo {Celsius}°C para Fahrenheit: {Fahrenheit}°F")
````

## Questão 10 - Preço com IVA e desconto  *(2,2 valores)*

Uma loja de informática quer calcular o preço final de um artigo. Peça:

- o **preço sem IVA** (em euros);
- a **percentagem de desconto** a aplicar (ex.: `10`).

Calcule e mostre, sabendo que o **IVA é de 23%**:

a) o valor do **IVA**;
b) o **preço com IVA**;
c) o valor do **desconto** (aplicado sobre o preço com IVA);
d) o **preço final** a pagar.

````python
preco_sem_iva = float(input("Valor sem IVA: "))
desconto = float(input("Valor de desconto: "))
iva = 0.23
preco_final = preco_sem_iva - desconto % 0.23 

print(f"Preço sem IVA:{preco_sem_iva} Desconto: {desconto} IVA:{iva} Preço:{preco_final} final a pagar ")
````

## Questão 11 - Orçamento de montagem de um PC  *(5,0 valores)*

Numa loja de assistência informática, um técnico prepara o orçamento da montagem de um computador. Escreva **um único programa** que peça os seguintes dados:

- nome do cliente;
- preço da **motherboard** (€);
- preço do **processador** (€);
- preço da **memória RAM** (€);
- preço do **disco SSD** (€);
- valor da **mão de obra** de montagem (€);
- **quantia em dinheiro** entregue pelo cliente para pagamento (€).

O programa deve calcular e mostrar, com mensagens claras e identificadas:

a) o **subtotal dos componentes** (sem mão de obra);
b) o **total dos componentes mais a mão de obra** (valor sem IVA);
c) o valor do **IVA (23%)** sobre o total anterior;
d) o **total a pagar** (com IVA);
e) o **troco** a devolver ao cliente (dinheiro entregue − total a pagar).

**Sugestão de apresentação final:**

```
===== ORCAMENTO =====
Cliente: <nome>
Subtotal componentes: <valor> EUR
Total sem IVA: <valor> EUR
IVA (23%): <valor> EUR
TOTAL A PAGAR: <valor> EUR
Troco: <valor> EUR
```

> *Nota: neste exercício não é necessário verificar se o dinheiro entregue chega — basta calcular a diferença.*

````python
nome_cliente = input("Qual o nome do Cliente? ")
motherboard = float(input("Preço da motherboard: "))
processador = float(input("Preço do processador: "))
ram = float(input("Preço da mémoria RAM: "))
ssd = float(input("Preço do ssd: "))
MaoDeObra = float(input("Valor da Mão de Obra: "))
pagamento = float(input("Quantia em dinheiro: "))
Subtotal = motherboard + processador + ram + ssd 
total_sem_iva = Subtotal + MaoDeObra 
iva = 0.23 
total_a_pagar =  iva + total_sem_iva
troco = pagamento - total_a_pagar

print("===== ORÇAMENTO =====")
print("Cliente: " + nome_cliente)
print(f"Subtotal componentes: {Subtotal} EUR")
print(f"Total sem IVA: {total_sem_iva} EUR")
print(f"IVA (23%): {iva} EUR")
print(f"TOTAL A PAGAR: {total_a_pagar} EUR")
print(f"Troco: {troco} EUR")

````
