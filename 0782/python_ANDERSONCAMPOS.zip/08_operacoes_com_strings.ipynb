## 2. Strings em Python — revisão

Uma string é uma sequência de caracteres delimitada por aspas. Tens **três formas** de a delimitar:

````python
# Aspas duplas
frase1 = "Bom dia"

# Aspas simples (equivalentes)
frase2 = 'Bom dia'

# Aspas triplas — para strings com várias linhas
frase3 = """Uma string que
ocupa várias linhas."""

print(frase1)
print(frase2)
print(frase3)
````

### 2.2 Concatenação e repetição

Já vistos no Manual 5 — recordo apenas:

````python
# Concatenação com +
print("bom" + " " + "dia")

# Repetição com *
print("-" * 30)
````

### 2.3 Imutabilidade

**Strings em Python são imutáveis.** Depois de criadas, não podes alterar nenhum dos seus caracteres — tens de criar uma string nova.

A célula seguinte vai dar **erro** propositadamente, para veres a mensagem:

````python
nome = "maria"
nome[0] = "M"   # Vai lançar TypeError
````

Para obter o efeito de "alterar", criamos uma string nova:

````python
nome = "maria"
nome = "M" + nome[1:]    # cria string nova
print(nome)

# Alternativa mais elegante:
nome = "maria"
nome = nome.capitalize()
print(nome)
````

## 3. Indexação

Cada caráter ocupa uma posição numerada. Acedes a um caráter com `string[indice]`.

### 3.1 Índices positivos

Contam-se da esquerda para a direita, **começando em 0** (não em 1!).

```
 P y t h o n
 0 1 2 3 4 5
```

````python
frase = "Python"

print(frase[0])   # 'P'
print(frase[1])   # 'y'
print(frase[2])   # 't'
print(frase[5])   # 'n'
````

### 3.2 Índices negativos

Contam a partir do fim: `-1` é o último, `-2` o penúltimo, etc.

```
  P  y  t  h  o  n
 -6 -5 -4 -3 -2 -1
```

````python
frase = "Python"

print(frase[-1])   # 'n'
print(frase[-2])   # 'o'
print(frase[-6])   # 'P'
````

### 3.3 Erros de índice

Aceder a uma posição que não existe gera `IndexError`:

````python
frase = "Python"
frase[10]   # IndexError: string index out of range
````

## 4. Slicing — extrair fatias

### 4.1 Sintaxe `[início:fim]`

Extrai uma fatia da string. **`início` é inclusivo, `fim` é exclusivo.**

Isto significa que `frase[0:3]` devolve 3 caracteres — índices 0, 1 e 2.

````python
frase = "Python"
#         0 1 2 3 4 5

print(frase[0:3])   # 'Pyt'
print(frase[2:5])   # 'tho'
print(frase[1:4])   # 'yth'
````

### 4.2 Omitir início ou fim

- Omitir o início → começa no princípio.
- Omitir o fim → vai até ao fim.
- Omitir ambos → cópia da string.

````python
frase = "Python"

print(frase[:3])    # 'Pyt'    — do início ao índice 3
print(frase[3:])    # 'hon'    — do índice 3 ao fim
print(frase[:])     # 'Python' — cópia completa
````

````python
# Índices negativos também funcionam em slicing
ficheiro = "relatorio_2026.pdf"

print(ficheiro[-3:])      # 'pdf'             — últimas 3 letras (extensão)
print(ficheiro[:-4])      # 'relatorio_2026'  — tudo menos os últimos 4
print(ficheiro[-8:-4])    # '2026'
````

### 4.3 Slicing com passo `[início:fim:passo]`

O terceiro parâmetro indica de quantos em quantos saltar (por defeito, 1).

````python
frase = "abcdefghij"

print(frase[0:10:2])   # 'acegi'   — de 2 em 2
print(frase[::2])      # 'acegi'   — idem, mais conciso
print(frase[::3])      # 'adgj'    — de 3 em 3
print(frase[1::2])     # 'bdfhj'   — do índice 1, de 2 em 2
````

### 4.4 Inverter uma string

Passo `-1` percorre a string ao contrário. É a forma idiomática de inverter:

````python
print("Python"[::-1])              # 'nohtyP'
print("socorram-me"[::-1])         # 'em-marrocos' (quase!)
````

## 5. Comprimento e contagem

- `len(s)` devolve o número de caracteres.
- `s.count(sub)` conta ocorrências de uma sub-string.

````python
# len() conta todos os caracteres — letras, dígitos, espaços, pontuação
print(len("Python"))      # 6
print(len(""))             # 0
print(len("   olá!  "))    # 9
````

````python
frase = "banana"

print(frase.count("a"))    # 3
print(frase.count("an"))   # 2
print(frase.count("x"))    # 0

# Intervalo opcional — procurar só numa parte
print("banana".count("a", 0, 3))    # 1 (só nos primeiros 3 caracteres)
````

## 6. Métodos de transformação

Lembra-te: **cada método devolve uma string nova** — a original mantém-se intacta.

### 6.1 Mudança de capitalização

| Método | Efeito |
|---|---|
| `.upper()` | Tudo em maiúsculas |
| `.lower()` | Tudo em minúsculas |
| `.title()` | Maiúscula em cada palavra |
| `.capitalize()` | Maiúscula só na primeira letra |
| `.swapcase()` | Inverte maiúsculas/minúsculas |

````python
nome = "maria"

print(nome.upper())        # 'MARIA'
print(nome.capitalize())   # 'Maria'
print(nome)                # 'maria' — a original mantém-se!
````

````python
print("bom dia".title())          # 'Bom Dia'
print("Bom Dia".swapcase())       # 'bOM dIA'

# Encadeamento (chaining) — aplicar vários métodos em sequência
print("INFORMATICA".lower().capitalize())   # 'Informatica'
````

### 6.2 Aparar espaços

- `.strip()` — remove espaços (e `\n`, `\t`) do início e do fim.
- `.lstrip()` — só do início.
- `.rstrip()` — só do fim.

````python
texto = "   ola   "

print(repr(texto.strip()))     # 'ola'
print(repr(texto.lstrip()))    # 'ola   '
print(repr(texto.rstrip()))    # '   ola'

# Nota: usei repr() para se verem os espaços nos resultados
````

````python
# Argumento opcional — quais caracteres remover
print("---ola---".strip("-"))                  # 'ola'
print("www.exemplo.pt/".strip("/"))            # 'www.exemplo.pt'
````

### 6.3 Substituir texto

`.replace(antigo, novo)` substitui **todas** as ocorrências.

````python
print("O João tem um carro".replace("João", "Pedro"))
print("banana".replace("a", "o"))
print("   espaços   ".replace(" ", ""))

# Terceiro argumento opcional — limitar o nº de substituições
print("banana".replace("a", "o", 2))    # 'bonona' — só as 2 primeiras
````

## 7. Métodos de pesquisa

### 7.1 `.find()` — devolve o índice da primeira ocorrência

Se não encontrar, devolve `-1`.

````python
print("banana".find("na"))    # 2
print("banana".find("x"))     # -1 (não encontrado)

print("O ficheiro tem extensao .pdf".find("."))    # 24
````

### 7.2 `.startswith()` e `.endswith()`

Verificam se a string começa/termina com determinado prefixo/sufixo. Devolvem `True` ou `False`.

````python
print("relatorio.pdf".endswith(".pdf"))           # True
print("relatorio.pdf".endswith(".docx"))          # False

print("https://exemplo.pt".startswith("https://"))   # True
print("https://exemplo.pt".startswith("http"))       # True
````

### 7.3 Operador `in`

Para apenas saber se uma sub-string existe (sem precisar do índice). É a forma idiomática em Python.

````python
print("py" in "python")                  # True
print("java" in "python")                # False
print("@" in "maria@exemplo.pt")         # True

# Negação: "not in"
print("z" not in "python")               # True
````

## 8. Dividir e juntar strings

### 8.1 `.split()` — dividir em partes

Devolve uma **lista** de strings.

````python
print("bom dia bom dia".split(" "))             # ['bom', 'dia', 'bom', 'dia']
print("maca,laranja,banana,uva".split(","))     # ['maca', 'laranja', 'banana', 'uva']
print("2026-04-26".split("-"))                  # ['2026', '04', '26']

# Sem argumento — divide por qualquer espaço em branco (vários espaços = um só)
print("bom    dia   tarde".split())             # ['bom', 'dia', 'tarde']
````

### 8.2 `.join()` — juntar partes

É chamado a partir do **separador**, e recebe uma lista para juntar.

````python
print("-".join(["2026", "04", "26"]))                # '2026-04-26'
print(", ".join(["maca", "laranja", "banana"]))      # 'maca, laranja, banana'
print("".join(["P", "y", "t", "h", "o", "n"]))       # 'Python'
````

### Combinar `split` e `join` — substituições poderosas

Padrão idiomático para reduzir múltiplos espaços a um só:

````python
frase = "bom    dia    tarde"

# 1. .split() (sem argumento) parte em palavras, descartando espaços extra
# 2. " ".join(...) junta com um único espaço
limpa = " ".join(frase.split())

print(limpa)    # 'bom dia tarde'
````

## 9. Verificar tipo de conteúdo

Família de métodos `is..()` que devolvem `True` ou `False`.

| Método | Devolve True se a string contém apenas... |
|---|---|
| `.isdigit()` | dígitos (0-9) |
| `.isalpha()` | letras |
| `.isalnum()` | letras ou dígitos |
| `.isspace()` | espaços, tabulações, quebras de linha |
| `.isupper()` | letras em maiúsculas |
| `.islower()` | letras em minúsculas |

⚠️ **String vazia devolve `False` em todos.**

````python
print("12345".isdigit())     # True
print("123a".isdigit())      # False
print("".isdigit())          # False — string vazia!

print("Maria".isalpha())     # True
print("Maria 25".isalpha())  # False — tem espaço e dígitos

print("PYTHON".isupper())    # True
print("Python".isupper())    # False

# Atenção: não aceita negativos nem decimais
print("-123".isdigit())      # False — o '-' não é dígito
print("3.14".isdigit())      # False — o '.' não é dígito
````

### Aplicação — validar uma idade introduzida

Esta abordagem (verificar antes de converter) evita erros de conversão sem precisar de `try/except`:

````python
# Para testar no notebook, substituímos input() por um valor fixo
# Em código real, usar:  entrada = input("Idade: ").strip()
entrada = "25"

if entrada.isdigit():
    idade = int(entrada)
    print(f"Idade registada: {idade}")
else:
    print("Por favor, insere apenas dígitos.")
````

## 10. Aplicações práticas

### 10.1 Validar dados de entrada: email

Juntamos várias técnicas:

````python
# Para testar, substituímos input() por um valor fixo
# Em código real: email = input("Email: ").strip().lower()
email = "Maria.Silva@Exemplo.PT".strip().lower()

if "@" not in email:
    print("Falta o @ no email")
elif email.count("@") > 1:
    print("O email tem mais do que um @")
elif "." not in email:
    print("Falta um ponto no email")
elif email.startswith("@") or email.endswith("@"):
    print("O @ está numa posição inválida")
else:
    print(f"Email válido: {email}")
````

### 10.2 Limpeza de texto

Recebemos uma string "suja" e normalizamos para análise:

````python
texto = "   Bom DIA, como estás??!!   "

# 1. Aparar e converter para minúsculas
limpo = texto.strip().lower()

# 2. Substituir pontuação por espaço
for sinal in ["?", "!", ",", "."]:
    limpo = limpo.replace(sinal, " ")

# 3. Reduzir múltiplos espaços a um só
limpo = " ".join(limpo.split())

print(f"Texto limpo: '{limpo}'")
````

### 10.3 Análise de uma frase

Contar palavras, caracteres (sem espaços) e a palavra mais longa:

````python
# Para testar: substitui input() por um valor fixo
# Em código real: frase = input("Escreve uma frase: ").strip()
frase = "A programacao em Python e divertida e facil"

palavras = frase.split()
n_palavras = len(palavras)
n_caracteres = len(frase.replace(" ", ""))

# Encontrar a palavra mais longa
mais_longa = ""
for palavra in palavras:
    if len(palavra) > len(mais_longa):
        mais_longa = palavra

print(f"Número de palavras:  {n_palavras}")
print(f"Número de caracteres (sem espaços): {n_caracteres}")
print(f"Palavra mais longa:  '{mais_longa}' ({len(mais_longa)} letras)")
````
