## 1. Operador, operando, expressão

Considera a expressão:

```
5  +  3
│   │   │
│   │   └── operando
│   └────── operador
└────────── operando
```

Resultado da avaliação: **8**

- **Operador** — símbolo que representa uma operação a fazer sobre um ou mais valores.
- **Operando** — o(s) valor(es) sobre o(s) qual(is) o operador atua.
- **Expressão** — qualquer combinação de valores, variáveis e operadores que possa ser avaliada para produzir um resultado.

````python
# Uma expressão simples
5 + 3
````

| Tipo | Atua sobre | Exemplo |

| **Unário** | um único operando | `-5` (o menos a negar 5) |
| **Binário** | dois operandos | `5 + 3` (a maioria deste manual) |

````python
# Unário: o menos nega um único valor
print(-5)

# Binário: soma dois operandos
print(5 + 3)
````

## 3. Operadores aritméticos básicos

Como na matemática que já conheces:

| Operador | Operação | Exemplo | Resultado |
|----------|----------|---------|-----------|
| `+` | Soma | `10 + 3` | `13` |
| `-` | Subtração | `10 - 3` | `7` |
| `*` | Multiplicação | `10 * 3` | `30` |
| `/` | Divisão (real) | `10 / 3` | `3.333…` |

> ⚠️ **Em Python 3, `/` dá SEMPRE um `float`** — mesmo quando a divisão é exata. `10 / 2` dá `5.0`, não `5`.

````python
print(10 + 3) 
print(10 - 3) 
print(10 * 3) 
print(10 / 3) 

# Atenção: divisão dá sempre float
print(10 / 2)   # 5.0, não 5
print(9 / 3)    # 3.0

# Confirma o tipo
print(type(10 / 2))
````

## 4. Divisão inteira `//` e módulo `%`

Dois operadores muito úteis:

- `//` — **divisão inteira**: descarta a parte decimal (quantas vezes cabe).
- `%` — **módulo**: devolve o resto da divisão (o que sobra).

**Intuição:** `//` → *quantas vezes cabe*; `%` → *o que sobra*.

````python
# Divisão inteira //
print(17 // 5)   # 3
print(20 // 4)   # 5
print(10 // 3)   # 3

print("---")

# Módulo % (resto)
print(17 % 5)    # 2  -> resto
print(20 % 4)    # 0  -> múltiplo!
print(10 % 3)    # 1
````

## 5. Para que serve o operador `%`

Padrões muito usados na prática:

- **Par ou ímpar** — `numero % 2 == 0` → um número é par se o resto da divisão por 2 for zero.
- **Múltiplo de N** — `x % n == 0` → 15 é múltiplo de 5 porque `15 % 5 == 0`.
- **Horas e minutos** — `seg // 60` dá os minutos, `seg % 60` dá os segundos restantes.

Este padrão será o nosso *pão diário* em ciclos e decisões.

````python
numero = 8
print("8 é par?", numero % 2 == 0)

numero = 7
print("7 é par?", numero % 2 == 0)

print("15 é múltiplo de 5?", 15 % 5 == 0)
````

## 6. Potência `**` e unário menos `-`

### Potência `**`

Elevar a `0.5` é o mesmo que tirar a **raiz quadrada** — propriedade matemática útil que evita importar `math`.

### Unário menos `-`

Inverte o sinal. Existe também o unário `+`, mas só por simetria (não muda o valor).

````python
# Potência
print(2 ** 3)      # 8
print(5 ** 2)      # 25
print(10 ** 6)     # 1000000
print(16 ** 0.5)   # 4.0  -> raiz quadrada

print("---")

# Unário menos
print(-5)
x = 7
print(-x)          # -7
print(-(-3))       # 3
````

## 7. Operadores de atribuição composta

Atualizar uma variável com base no seu próprio valor é tão comum que tem **forma curta**:

| Operador | Equivale a | Exemplo |
|----------|-----------|---------|
| `+=` | `x = x + valor` | `x += 5` |
| `-=` | `x = x - valor` | `x -= 3` |
| `*=` | `x = x * valor` | `x *= 2` |
| `/=` | `x = x / valor` | `x /= 4` |
| `//=` | `x = x // valor` | `x //= 4` |
| `%=` | `x = x % valor` | `x %= 3` |
| `**=` | `x = x ** valor` | `x **= 2` |

````python
x = 10
x += 5      # x = x + 5
print(x)    # 15

x *= 2      # x = x * 2
print(x)    # 30

x //= 4     # x = x // 4
print(x)    # 7

x **= 2     # x = x ** 2
print(x)    # 49
````

## 8. Python NÃO tem `++` nem `--`

Diferença em relação a C, C++, Java e JavaScript:

| ✗ Não existe em Python | ✓ Usa-se em alternativa |
|------------------------|-------------------------|
| `x++`, `x--`, `++x`, `--x` | `x += 1`, `x -= 1` |

**Porquê?** Escolha deliberada. O `++` é uma fonte clássica de *bugs* subtis — a diferença entre `x++` (pós-incremento) e `++x` (pré-incremento) confunde toda a gente. Python prefere o caminho explícito.

````python
x = 5
x += 1      # em vez de x++
print(x)    # 6

x -= 1      # em vez de x--
print(x)    # 5

# A linha seguinte daria SyntaxError em Python — experimenta descomentar:
# x++
````

## 9. Operador `+` com *strings* · concatenação

Aplicado a duas *strings*, o `+` devolve a junção das duas — **sem espaço**.

> ⚠️ **Não se somam *strings* com números.** `"123" + 4` dá `TypeError`. Solução: converter com `str(4)` — ou, melhor ainda, usar *f-strings* (Manual 6).

````python
print("bom" + "dia")          # 'bomdia'  -> sem espaço
print("bom" + " " + "dia")     # 'bom dia' -> espaço explícito

nome = "Maria"
print("Ola, " + nome + "!")    # 'Ola, Maria!'
````

````python
# Isto dá erro — descomenta para veres a mensagem TypeError:
# print("123" + 4)

# Solução: converter o número para string
print("123" + str(4))          # '1234'
````

## 11. Precedência de operadores

Quando uma expressão tem vários operadores, o Python segue regras precisas — as **regras de precedência**, quase iguais às da matemática.

Exemplo passo a passo de `2 + 3 * 4`:
1. **Primeiro o `*`** (maior precedência que a soma): `3 * 4 = 12`
2. **Depois o `+`**: `2 + 12 = 14`

Resultado: **14** (não 20!). Com parêntesis muda tudo: `(2 + 3) * 4 → 20`.

````python
print(2 + 3 * 4)      # 14  -> multiplica primeiro
print((2 + 3) * 4)    # 20  -> parêntesis forçam a soma primeiro
````

### Tabela de precedência (do maior para o menor)

| Prioridade | Operadores | Descrição |
|------------|-----------|-----------|
| 1 (mais alto) | `( )` | Parêntesis |
| 2 | `**` | Potência |
| 3 | `+x`, `-x` | Operadores unários |
| 4 | `* / // %` | Multiplicação, divisão, divisão inteira, módulo |
| 5 (mais baixo) | `+ -` | Soma, subtração (e concatenação de *strings*) |

````python
# Aplicando a tabela: 2 ** 3 + 4 * 5
# (1) 2**3 = 8   (2) 4*5 = 20   (3) 8 + 20 = 28
print(2 ** 3 + 4 * 5)   # 28
````

## 12. Associatividade

Quando os operadores têm a **mesma prioridade**:

- **Regra geral:** da **esquerda para a direita**.
- **Exceção:** `**` associa-se à **direita**.

````python
# Regra geral: esquerda -> direita
# 100 / 5 / 2  =  (100/5)/2  =  20/2  =  10.0
print(100 / 5 / 2)   # 10.0
````

````python
# Exceção: ** associa-se à direita
# 2 ** 3 ** 2  =  2 ** (3 ** 2)  =  2 ** 9  =  512
print(2 ** 3 ** 2)     # 512
print((2 ** 3) ** 2)   # 64  -> com parêntesis explícitos
````

## 13. Quando usar parêntesis

**Regra de ouro: na dúvida, usa-os.** Mesmo quando matematicamente desnecessários, tornam a intenção clara para quem vai ler o código depois — incluindo tu próprio.

````python
a, b, c = 10, 20, 30

# ✗ Calcula mal a média: faz c/3 primeiro
media_errada = a + b + c / 3
print(media_errada)        # 40.0  -> a + b + (c/3)

# ✓ Calcula bem a média
media_certa = (a + b + c) / 3
print(media_certa)         # 20.0
````

## 14. Casos práticos

### 14.1 · Cálculo de IVA

Dado o preço sem IVA, calcular o preço final com IVA à taxa normal (23%).

````python
TAXA_IVA = 0.23
preco_sem_iva = 100.00

# Versão explícita
valor_iva = preco_sem_iva * TAXA_IVA
preco_com_iva = preco_sem_iva + valor_iva

print("Preco sem IVA:", preco_sem_iva)
print("Valor do IVA: ", valor_iva)
print("Preco com IVA:", preco_com_iva)

# Versão compacta (numa só linha)
preco_com_iva = preco_sem_iva * (1 + TAXA_IVA)
print("Versao compacta:", preco_com_iva)
````

### 14.2 · Conversão de unidades

Repara na **precedência em ação**: em `temp_c * 9 / 5 + 32`, o `*` e o `/` aplicam-se da esquerda para a direita e o `+` fica para o fim — exatamente o que queremos.

````python
# Quilómetros -> metros
distancia_km = 5.5
distancia_m = distancia_km * 1000
print(distancia_km, "km =", distancia_m, "m")

# Celsius -> Fahrenheit
temp_c = 25
temp_f = temp_c * 9 / 5 + 32
print(temp_c, "C =", temp_f, "F")
````

### 14.3 · Horas, minutos e segundos — o clássico do `//` e `%`

Dada uma duração total em segundos, decompor em horas, minutos e segundos. Aplica este padrão a cêntimos/euros, dezenas/centenas/milhares, paginação…

````python
total_segundos = 3725

horas = total_segundos // 3600   # quantas horas inteiras
resto = total_segundos % 3600    # o que sobra
minutos = resto // 60            # minutos nesse resto
segundos = resto % 60            # segundos finais

print(total_segundos, "segundos =", horas, "h", minutos, "m", segundos, "s")
````

## 15. Resumo do tópico

- **Operador** = símbolo; **operando** = valor sobre o qual atua; **expressão** = combinação avaliável.
- Aritméticos básicos: `+`, `-`, `*`, `/`. Em Python 3, `/` dá sempre `float`.
- `//` é a divisão inteira (descarta a decimal); `%` é o módulo (resto da divisão).
- `**` é a potência; `** 0.5` calcula a raiz quadrada.
- Atribuição composta: `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, `**=` abreviam `x = x op valor`.
- Python **NÃO** tem `++` nem `--` — usa-se `x += 1` e `x -= 1`.
- Para *strings*: `+` concatena (sem espaço), `*` repete. Não se soma *string* com número diretamente.
- Precedência: `( )` > `**` > unários > `* / // %` > `+ -`. O `**` é associativo à direita; os outros à esquerda.
- **Na dúvida, usa parêntesis.** Tornam a intenção clara e evitam *bugs*.
