## 2. A função `print()` em detalhe

### 2.1. Sintaxe e funcionamento

É a função mais usada do Python e a forma *standard* de mostrar texto e valores no ecrã. Converte automaticamente os valores em texto e termina com uma mudança de linha (o cursor passa para a linha seguinte).

```python
print("Bom dia")
print(42)
print(3.14)
print(True)
```

### 2.2. Imprimir vários valores

A `print()` aceita vários argumentos separados por vírgula. Imprime-os todos, separando-os com um espaço por omissão. Podemos misturar *strings*, números e variáveis na mesma chamada.

```python
nome = "Maria"
idade = 25
print("O nome e", nome, "e a idade e", idade)
```

### 2.3. O parâmetro `sep`

O separador por defeito é um espaço, mas pode alterar-se com `sep`.

```python
print("a", "b", "c")
print("a", "b", "c", sep="-")
print("a", "b", "c", sep=", ")
print("a", "b", "c", sep="")

# Útil para construir datas ou caminhos
print(2026, 4, 26, sep="/")
print("home", "aluno", "documentos", sep="/")
```

### 2.4. O parâmetro `end`

Por defeito, cada `print()` termina com uma mudança de linha (`\n`). Com `end` mudamos esse comportamento — útil para manter texto na mesma linha.

```python
# Por defeito: cada print numa linha
print("primeiro")
print("segundo")

print("---")

# Com end=" ": continua na mesma linha
print("primeiro", end=" ")
print("segundo")
```

```python
# sep e end combinados
print("a", "b", "c", sep="-", end="!\n")
```

## 3. Sequências de escape

Há carateres que não se conseguem escrever diretamente dentro de uma *string* (uma aspa dentro de aspas, uma mudança de linha…). A solução são as **sequências de escape**: barra invertida (`\`) seguida de uma letra ou símbolo.

| Sequência | Significado | Exemplo |
|-----------|-------------|---------|
| `\n` | Mudança de linha | `"linha1\nlinha2"` |
| `\t` | Tabulação | `"col1\tcol2"` |
| `\\` | Barra invertida literal | `"C:\\dados"` |
| `\"` | Aspas duplas dentro de `""` | `"Disse \"ola\""` |
| `\'` | Aspas simples dentro de `''` | `'L\'aluno'` |
| `\r` | Retorno de carro (raro) | `"abc\rxyz"` |
| `\b` | Backspace (raro) | `"abc\b"` |

```python
print("linha1\nlinha2\nlinha3")
print("---")
print("Nome\tIdade\tCidade")
print("Joao\t25\tSantarem")
print("---")
print("O caminho e C:\\Users\\aluno")
print("Ela disse \"bom dia\" ao chegar")
```

> **Aspas dentro de aspas — alternativa elegante.** Em vez de escapar as aspas, delimita a *string* com o outro tipo de aspas. As duas formas abaixo dão o mesmo resultado; a segunda costuma ser mais legível.

```python
print("Ela disse \"bom dia\"")   # com escape
print('Ela disse "bom dia"')      # com aspas simples — mais legível
```

### 3.2. *Strings* cruas (raw strings)

Ao escrever caminhos de Windows ou expressões regulares (cheios de barras invertidas), as sequências de escape tornam-se um inferno. O prefixo `r` cria uma **raw string**, onde a barra invertida é tratada literalmente.

```python
# A forma idiomática: prefixo r
caminho = r"C:\Users\aluno\projeto"
print(caminho)

# Sem r, teríamos de duplicar todas as barras:
caminho2 = "C:\\Users\\aluno\\projeto"
print(caminho2)
```

## 4. f-strings — a forma moderna de formatar

### 4.1. Sintaxe básica

Uma f-string é uma *string* precedida da letra `f`. Dentro dela, podes incluir nomes de variáveis e expressões entre chavetas `{}`, que são automaticamente substituídos pelo seu valor. Sem conversões manuais, sem concatenação, sem vírgulas.

```python
nome = "Maria"
idade = 25
print(f"O nome e {nome} e a idade e {idade}")
```

> **Não esquecer o `f`!** Sem o `f` no início, as chavetas são tratadas como texto normal e nada é substituído.

```python
nome = "Ana"
print("Ola, {nome}")    # falta o f -> imprime literalmente {nome}
print(f"Ola, {nome}")   # com f -> substitui pelo valor
```

### 4.2. Expressões dentro de f-strings

Dentro das chavetas pode estar **qualquer expressão Python** — não só nomes de variáveis. Operações, chamadas a funções, comparações: tudo é avaliado e o resultado entra na *string*.

```python
nome = "Maria"
preco = 12.50
quantidade = 4

print(f"Total: {preco * quantidade} euros")
print(f"O dobro de {preco} e {preco * 2}")
print(f"Tem desconto? {preco > 10}")
print(f"Em maiusculas: {nome.upper()}")
```

### 4.3. Especificadores de formato

A grande superpotência das f-strings: instruções sobre **como** o valor aparece, indicadas depois de dois pontos `:` dentro das chavetas — `f"{valor:formato}"`. Para mostrar um número com duas casas decimais usamos `.2f` (`f` de float, `.2` para duas casas).

```python
preco = 12.5
print(f"Preco: {preco} euros")       # sem especificador -> 12.5
print(f"Preco: {preco:.2f} euros")   # com .2f -> 12.50
```

### 4.4. Largura, alinhamento e preenchimento

Sintaxe geral: `{valor:[fill][align][width]}`

- **fill** — caráter de preenchimento (omitido = espaço)
- **align** — `<` (esquerda), `>` (direita), `^` (centro)
- **width** — número mínimo de caracteres

```python
nome = "Ana"
print(f"|{nome:<10}|")    # esquerda, largura 10
print(f"|{nome:>10}|")    # direita, largura 10
print(f"|{nome:^10}|")    # centrado, largura 10
print(f"|{nome:*^10}|")   # centrado, preenchido com *
```

### 4.5. Casas decimais e separadores

| Formato | Significado | Exemplo | Resultado |
|---------|-------------|---------|-----------|
| `.2f` | Float com 2 casas | `f"{3.14159:.2f}"` | `3.14` |
| `.0f` | Float sem casas | `f"{3.7:.0f}"` | `4` (arredonda) |
| `,` | Separador de milhares | `f"{1000000:,}"` | `1,000,000` |
| `_` | Sublinhado como separador | `f"{1000000:_}"` | `1_000_000` |
| `,.2f` | Combinado | `f"{1234.5:,.2f}"` | `1,234.50` |
| `%` | Percentagem | `f"{0.42:.1%}"` | `42.0%` |
| `e` | Notação científica | `f"{1234.5:.2e}"` | `1.23e+03` |

```python
populacao = 10290000
print(f"Populacao: {populacao:,} habitantes")

taxa = 0.235
print(f"Taxa de IVA: {taxa:.1%}")

total = 1234.567
print(f"Total: {total:,.2f} euros")
```

> **Cuidado — separador de milhares à anglo-saxónica.** O `,` usa vírgula para milhares e ponto para decimais (o contrário do português). Para utilizadores portugueses pode ser preciso uma substituição manual, ou recorrer ao módulo `locale` (fica para mais tarde).

```python
total = 1234.567
valor_pt = f"{total:,.2f}".replace(",", " ").replace(".", ",")
print(valor_pt)   # 1 234,57
```

### 4.6. Formatos numéricos especiais

Para inteiros há especificadores para mostrar noutras bases — úteis em redes, criptografia e baixo nível.

```python
n = 255
print(f"Decimal: {n}")
print(f"Binario: {n:b}")
print(f"Octal: {n:o}")
print(f"Hexadecimal: {n:x}")
print(f"Hexa em maiusculas: {n:X}")
```

**Truque: f-string com debug (Python 3.8+).** Um `=` depois do nome da variável mostra o nome E o valor — excelente para depuração rápida.

```python
idade = 25
nome = "Ana"
print(f"{idade=}, {nome=}")
```

## 5. Outras formas de formatar (legado)

Antes das f-strings havia (e ainda há) duas outras formas. **Não as uses no teu código** — mas reconhece-as quando aparecerem em código antigo.

### 5.1. Método `.format()`

Pontos de inserção com `{}`; os valores passam-se como argumentos. Os especificadores depois de `:` funcionam **exatamente** como nas f-strings.

```python
nome = "Maria"
idade = 25

# Posicional
print("O nome e {} e a idade e {}".format(nome, idade))
# Com indices (permite reutilizar)
print("{0} tem {1} anos. {0} mora em Lisboa.".format(nome, idade))
# Com nomes
print("{n} tem {i} anos".format(n=nome, i=idade))
# Com especificadores
print("Preco: {:.2f} euros".format(12.5))
```

### 5.2. `%`-formatting

A forma mais antiga, herdada do C. Considerada legada. Marcadores comuns: `%s` (string), `%d` (inteiro), `%f` (float, com `.2f` para 2 casas), `%x` (hexadecimal).

> **Reconhecer, não usar.** Ainda é comum em *logs* (o módulo `logging` usa-o por razões históricas). Para o teu código, prefere f-strings.

```python
nome = "Maria"
idade = 25
preco = 12.5

print("O nome e %s e a idade e %d" % (nome, idade))
print("Preco: %.2f euros" % preco)
```

## 6. Carateres especiais e Unicode

O Python 3 trabalha por defeito com **Unicode**. Podes usar acentos, `ç`, símbolos matemáticos, alfabetos não latinos e até emojis em *strings* sem qualquer configuração.

```python
print("Olá, mundo!")
print("Coração: \u2764")
print("\u03c0 = 3.14159")     # pi
print("日本語")
```

> **Porque evitamos acentos em código?** O Python aceita acentos em *strings* sem problema (e deves usá-los em texto para o utilizador). Mas é convenção mantê-los **longe** dos nomes de variáveis e funções — por portabilidade, interoperabilidade e simplicidade.

**Inserir carateres pelo código Unicode:** usa `\u` seguido de 4 dígitos hexadecimais.

```python
print("\u00e9")    # é
print("\u20ac")    # euro
print("\u2764")    # coração
```

## 7. Aplicações práticas

### 7.1. Tabela de produtos

Combinando largura e alinhamento, criamos tabelas bem alinhadas em texto puro — sem bibliotecas externas. Os produtos alinham à esquerda em 20 caracteres; os preços à direita em 12, com 2 casas decimais.

```python
produto1 = "Cafe";          preco1 = 0.80
produto2 = "Sandes mista";  preco2 = 3.50
produto3 = "Sumo natural";  preco3 = 2.75

print(f"{'Produto':<20}{'Preco (EUR)':>12}")
print("-" * 32)
print(f"{produto1:<20}{preco1:>12.2f}")
print(f"{produto2:<20}{preco2:>12.2f}")
print(f"{produto3:<20}{preco3:>12.2f}")
```

### 7.2. Recibo formatado

Juntamos tudo — `print()`, f-strings, alinhamento e operadores de *string* — para gerar um recibo completo, sem nenhuma biblioteca externa.

```python
TAXA_IVA = 0.23
loja = "Cafe Central"
data = "2026-04-26"

preco_cafe = 0.80;   qtd_cafe = 2
preco_sandes = 3.50; qtd_sandes = 1

subtotal = preco_cafe * qtd_cafe + preco_sandes * qtd_sandes
iva = subtotal * TAXA_IVA
total = subtotal + iva

linha = "=" * 40
print(linha)
print(f"{loja:^40}")
print(f"{'Data: ' + data:^40}")
print(linha)
print(f"{'Cafe (x' + str(qtd_cafe) + ')':<28}{preco_cafe * qtd_cafe:>12.2f}")
print(f"{'Sandes (x' + str(qtd_sandes) + ')':<28}{preco_sandes * qtd_sandes:>12.2f}")
print("-" * 40)
print(f"{'Subtotal':<28}{subtotal:>12.2f}")
print(f"{'IVA (23%)':<28}{iva:>12.2f}")
print(linha)
print(f"{'TOTAL':<28}{total:>12.2f}")
print(linha)
```

## 8. Resumo do tópico

- `print()` é a função padrão para mostrar valores no ecrã.
- Aceita vários argumentos separados por vírgula; modificáveis com `sep` (separador) e `end` (terminador, por defeito `\n`).
- Sequências de escape comuns: `\n`, `\t`, `\\`, `\"`, `\'`.

- Formas legadas `.format()` e `%`-formatting: reconhecer, não usar.
- Python 3 suporta Unicode nativamente.
- Combinando largura, alinhamento e formato criam-se tabelas e relatórios bem alinhados.
