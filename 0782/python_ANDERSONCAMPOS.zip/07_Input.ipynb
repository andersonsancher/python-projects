## 2. A função `input()`

### 2.1. Sintaxe básica

A função `input()` pede um valor ao utilizador e devolve aquilo que ele escreveu. Aceita opcionalmente uma mensagem (*prompt*) que aparece antes da escrita, indicando o que se espera.

````python
nome = input("Como te chamas? ")
print("Ola,", nome)
````

> **Pequeno detalhe da mensagem.** Costuma ser boa prática terminar a mensagem com um espaço, para o cursor não ficar colado à última letra:
> - `input("Idade?")` → `Idade?25`
> - `input("Idade? ")` → `Idade? 25`

### 2.2. `input()` devolve SEMPRE uma string

Este é o ponto mais importante do manual — e a fonte de erros para muitos iniciantes. Mesmo que o utilizador escreva um número, o Python recebe-o como **texto**.

````python
valor = input("Escreve um numero: ")
print(type(valor))   # <class 'str'> mesmo que escrevas 25
````

Se tentarmos fazer contas com a entrada **sem a converter**, obtemos um erro. Executa a célula seguinte e escreve um número — vais ver o `TypeError`:

```
TypeError: unsupported operand type(s) for -: 'int' and 'str'
```

O operador `-` não sabe subtrair uma *string* a um inteiro. A solução é **converter** antes de usar.

````python
# Esta célula dá erro de propósito — escreve um número e observa o TypeError.
# (Comenta as linhas se quiseres saltar.)
idade = input("Idade: ")
ano_nascimento = 2026 - idade   # ERRO! idade é uma string
````

> **Diferença para outras linguagens.** Em C, `scanf("%d", &n)` já entrega um inteiro porque dissemos que esperávamos `%d`. Em Python, `input()` não distingue: tudo o que vem do teclado é texto, e a conversão é responsabilidade do programador. É uma opção deliberada — torna o tipo previsível e a conversão explícita.

## 3. Conversão da entrada para o tipo certo

### 3.1. Para inteiro — `int()`

Combinamos `input()` e `int()` numa única linha — leitura aninhada que se lê *de dentro para fora*: `input(...)` devolve `"25"`; `int("25")` converte para `25`; `idade` fica com `25`.

````python
idade = int(input("Idade: "))
ano_nascimento = 2026 - idade
print("Nasceste em", ano_nascimento)
````

### 3.2. Para vírgula flutuante — `float()`

Para entradas que possam ter parte decimal.

````python
altura = float(input("Altura em metros: "))
peso = float(input("Peso em kg: "))
imc = peso / (altura ** 2)
print(f"O teu IMC e {imc:.2f}")
````

### 3.3. Aceitar vírgula como separador decimal

Em Portugal escreve-se `1,75`, mas o Python só aceita o ponto. Se o utilizador escrever `1,75`, o `float()` falha. Solução: substituir vírgulas por pontos **antes** de converter.

````python
# Passo a passo
entrada = input("Altura em metros: ")
entrada = entrada.replace(",", ".")
altura = float(entrada)
print("Altura registada:", altura)
````

````python
# Tudo encadeado numa só linha
altura = float(input("Altura: ").replace(",", "."))
print("Altura registada:", altura)
````

> **Pensar no utilizador.** Para programas usados por pessoas comuns, vale a pena absorver esta variação. É uma linha de código que evita erros frustrantes.

## 4. Validação básica de entrada

O utilizador é imprevisível: espaços a mais, maiúsculas onde esperávamos minúsculas, campos vazios.

### 4.1. Aparar espaços com `.strip()`

Remove espaços (e mudanças de linha) do início e do fim. É quase sempre boa ideia aplicá-lo ao que vem de `input()`.

````python
nome = input("Nome: ").strip()
print(f"Bom dia, {nome}!")
# Sem .strip(), "  Maria  " imprimiria "Bom dia,   Maria  !"
````

### 4.2. Aceitar maiúsculas e minúsculas — `.lower()` / `.upper()`

Com `.lower()`, tanto `"sim"`, `"Sim"`, `"SIM"` como `"  SIM  "` são tratados como `"sim"`.

> Nota: o `if`/`elif`/`else` aparece aqui pela primeira vez — vais aprofundá-lo na UFCD 0783.

````python
resposta = input("Queres continuar? (sim/nao): ").strip().lower()

if resposta == "sim":
    print("Vamos a isso!")
elif resposta == "nao":
    print("Ate a proxima.")
else:
    print("Resposta nao reconhecida.")
````

### 4.3. Detetar entradas vazias

Se o utilizador carregar Enter sem escrever nada, `input()` devolve uma *string* vazia (`""`).

````python
nome = input("Nome: ").strip()

if nome == "":
    print("Nao introduziste um nome.")
else:
    print(f"Ola, {nome}!")
````

````python
# Forma mais idiomática: uma string vazia é "falsa" em contexto booleano
nome = input("Nome: ").strip()

if not nome:
    print("Nao introduziste um nome.")
else:
    print(f"Ola, {nome}!")
````

> **E se o utilizador escrever "vinte" em vez de 20?** Aqui `int("vinte")` gera `ValueError` e o programa termina abruptamente. A forma correta de tratar isto é com `try`/`except` (UFCD 0785). Por agora, **assumimos que o utilizador escreve sempre o tipo de valor pedido**. Em código profissional, esta assunção tem de ser sempre questionada.

## 5. Programas interativos completos

Juntamos `input()`, `print()` e tudo o resto em três pequenos programas reais.

### 5.1. Calculadora de IMC

O IMC calcula-se dividindo o peso (kg) pelo quadrado da altura (m).

````python
print("=== Calculadora de IMC ===")
print()

nome = input("Como te chamas? ").strip()
altura = float(input("Altura (em metros): ").replace(",", "."))
peso = float(input("Peso (em kg): ").replace(",", "."))

imc = peso / (altura ** 2)

print()
print(f"Resumo de {nome}:")
print(f"  Altura: {altura} m")
print(f"  Peso:   {peso} kg")
print(f"  IMC:    {imc:.2f}")
````

### 5.2. Conversor de moedas

Converter um valor de euros para dólares, dada a taxa de câmbio.

````python
TAXA_EURO_DOLAR = 1.07

print("=== Conversor de Euros para Dolares ===")
valor_euros = float(input("Valor em euros: ").replace(",", "."))

valor_dolares = valor_euros * TAXA_EURO_DOLAR

print(f"{valor_euros:.2f} EUR = {valor_dolares:.2f} USD")
print(f"(taxa de cambio: 1 EUR = {TAXA_EURO_DOLAR} USD)")
````

### 5.3. Calculadora de gorjeta

Dado o valor da conta, a percentagem de gorjeta e o número de pessoas, calcular quanto deixar, o total e o valor por pessoa. Repara no alinhamento à direita (`>8.2f`) para as colunas de valores.

````python
print("=== Calculadora de Gorjeta ===")
print()

conta = float(input("Valor da conta: ").replace(",", "."))
percentagem = float(input("Percentagem de gorjeta (ex: 10): "))
pessoas = int(input("Numero de pessoas a dividir: "))

gorjeta = conta * percentagem / 100
total = conta + gorjeta
por_pessoa = total / pessoas

print()
print(f"Conta:           {conta:>8.2f} EUR")
print(f"Gorjeta ({percentagem:.0f}%):    {gorjeta:>8.2f} EUR")
print(f"Total:           {total:>8.2f} EUR")
print(f"Por pessoa ({pessoas}): {por_pessoa:>8.2f} EUR")
````

> **Já estás a fazer programas reais.** Estes três exemplos juntam tudo dos manuais 1 a 6: variáveis, tipos, conversões, operadores, f-strings e entrada interativa. Pequenos utilitários que pedem dados, calculam e apresentam resultados são exatamente o tipo de *scripts* que vais escrever a toda a hora como Técnico/a de Informática de Sistemas.

## 6. Boas práticas

- **Mensagens claras** — diz o que esperas, com unidade ou formato. `"Idade?"` é fraco; `"Idade em anos: "` é melhor; para formatos específicos, dá um exemplo: `"Data (AAAA-MM-DD): "`.
- **Aplica `.strip()` por defeito** — aceita os espaços extra como "erros" do utilizador que não devem fazer o programa falhar.
- **Converte logo para o tipo certo** — não transportes uma *string* pelo programa para converter mais à frente; é fonte de *bugs*.
- **Confirma o que o programa entendeu** — para entradas importantes, mostra ao utilizador o que foi registado, para ele notar erros.
- **Não assumas — antecipa erros** — verifica o básico: a entrada não está vazia? o número é positivo? está num intervalo razoável?

> *"O utilizador final está sempre a meio de outra coisa."* Quanto mais claro e tolerante for o programa, melhor a experiência.

## 7. Resumo do tópico

- `input("mensagem")` mostra a mensagem e devolve o que o utilizador escrever.
- `input()` devolve **sempre** uma *string* — mesmo que se escreva um número.
- Para usar como número, converte com `int()` ou `float()`.
- Para utilizadores portugueses, substitui vírgula por ponto antes do `float()`.
- `.strip()` apara espaços indesejados no início e no fim.
- `.lower()` permite comparar respostas (sim/nao) sem distinguir maiúsculas.
- Uma *string* vazia é "falsa" — testa com `if not entrada`.
- Combinar `input()`, conversões e f-strings permite construir programas interativos completos.
- Boas práticas: mensagens claras, `.strip()` por defeito, conversão imediata, confirmação.
- Tratamento de erros sofisticado (`try`/`except`) fica para a UFCD 0785.
