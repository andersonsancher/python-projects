## 2. Como é que o Python lê o teu código
Quando executas um ficheiro Python, o interpretador lê-o **de cima para baixo, linha a linha**. Cada linha é considerada uma instrução, e o Python tenta executá-la antes de passar à seguinte. Vê este pequeno exemplo:

````python
print("Bom dia")
print("Hoje vamos aprender Python")
print("Vai correr bem")
````

Cada instrução ocupa uma linha. Não é necessário terminar cada linha com ponto e vírgula (como em C ou em Java); o fim da linha já indica o fim da instrução.

## 3. Indentação como elemento sintático
### 3.1. Por que motivo a indentação importa
Indentar significa empurrar uma linha de código para a direita, deixando espaços em branco no início. Em Python, a indentação tem um papel **fundamental**: é ela que define o que pertence a cada bloco. Linhas indentadas ao mesmo nível pertencem ao mesmo bloco; uma linha menos indentada (mais à esquerda) sai do bloco anterior.

````python
idade = 20
if idade >= 18:
    print("És maior de idade")
    print("Podes votar")
print("Fim do programa")
````

As duas linhas com `print("És maior de idade")` e `print("Podes votar")` estão indentadas (quatro espaços) e pertencem ao bloco do `if`. A linha `print("Fim do programa")`, encostada à esquerda, fica fora do `if` e executa-se sempre.

> **Comparação com C/Java/JavaScript.** Em C o mesmo bloco usaria chavetas: `if (idade >= 18) { ...; }`.  
Em Python **não há chavetas: a indentação é a delimitação**. Não é só estética, é sintaxe.

### 3.2. Tabulações ou espaços?
É tecnicamente possível usar tabulações (Tab) ou espaços. A recomendação oficial (PEP 8) é clara: **quatro espaços por nível de indentação**. As IDEs modernas (VS Code, PyCharm) convertem o Tab em quatro espaços automaticamente.

> **Não misturar Tab com espaços.** O Python proíbe a mistura no mesmo ficheiro. Se o fizeres, recebes um `TabError` ou `IndentationError`.

### 3.3. Erros típicos de indentação
Os dois erros mais comuns no início:

**`IndentationError: expected an indented block`**: o Python esperava um bloco indentado a seguir a `if`, `for` ou `def`, mas não o encontrou:
```python
if idade >= 18:
print("Es maior de idade")  # ERRO: deveria estar indentado
```
Correção:
```python
if idade >= 18:
    print("Es maior de idade")
```

**`IndentationError: unexpected indent`**: uma linha está indentada sem motivo:
```python
print("Ola")
    print("Adeus")  # ERRO: indentacao sem motivo
```

> **Lê sempre as mensagens de erro.** Em regra são claras e indicam a linha do problema. Quase sempre a mensagem diz-te exatamente o que está mal.

## 4. Comentários
Comentários são linhas (ou partes de linhas) que o interpretador **ignora**. Servem para deixar notas para quem lê o código, incluindo tu próprio, daqui a uns meses.

### 4.1. Comentários de linha
Qualquer texto a seguir a um cardinal (`#`) é comentário.

````python
# Este e um comentario que ocupa uma linha inteira
idade = 18  # Este aparece no fim da linha de codigo

# Um exemplo mais util:
preco_unidade = 12.50  # Em euros, sem IVA
quantidade = 4
total = preco_unidade * quantidade  # Calcula o total
print(total)
````

### 4.2. Comentários multilinha
O Python não tem marcador específico para comentários de várias linhas. A solução habitual é `#` no início de cada linha. Existe também a alternativa das aspas triplas (uma string não atribuída a nenhuma variável):

````python
# Este e um comentario
# que ocupa varias linhas
# para descrever em detalhe o que o programa faz.

"""
Este e um comentario em aspas triplas.
Funciona em multiplas linhas.
Tipicamente usa-se no inicio dos modulos e funcoes.
"""
print("Programa a correr")
````

### 4.3. Boas práticas de comentários
- Comentários devem explicar o **porquê**, não o **quê**.
- Evita comentários a explicar o óbvio.
- Mantém os comentários atualizados (um comentário desatualizado engana quem o lê).

Comentário **inútil** vs **útil**:

````python
x = 10
x = x + 1  # incrementa x em 1          <- inutil (repete o obvio)
x = x + 1  # avanca para o proximo dia util do mes   <- util
print(x)
````

## 5. Separação em linhas e expressões
Cada instrução fica numa linha separada e não há ponto e vírgula no fim. Há, contudo, algumas situações a conhecer.

**Várias instruções na mesma linha** (legal mas desencorajado pela PEP 8):

````python
a = 1; b = 2; c = 3  # legal mas desencorajado
print(a, b, c)
````

**Instruções demasiado longas**: podem ser partidas com `\` ou, preferencialmente, envolvendo a expressão em parêntesis:

````python
primeiro_valor, segundo_valor, terceiro_valor = 1, 2, 3
quarto_valor, quinto_valor = 4, 5

# Com barra invertida:
total = primeiro_valor + segundo_valor + terceiro_valor + \
        quarto_valor + quinto_valor

# Com parentesis (forma preferida):
total = (primeiro_valor + segundo_valor + terceiro_valor +
         quarto_valor + quinto_valor)
print(total)
````

## 6. A convenção `if __name__ == "__main__"`
Em muitos exemplos vais ver esta estrutura no fim dos ficheiros. Qualquer ficheiro `.py` pode ser (a) executado diretamente como programa principal, ou (b) importado por outro ficheiro como biblioteca. A variável especial `__name__` permite ao ficheiro saber em que situação está:
- Executado diretamente → `__name__` contém `"__main__"`.
- Importado por outro → `__name__` contém o nome do ficheiro (sem `.py`).

Assim, o código dentro do bloco só corre quando o ficheiro é executado diretamente.

````python
# meu_programa.py

def calcular_total(preco, quantidade):
    return preco * quantidade

def cumprimentar(nome):
    print("Bom dia,", nome)

if __name__ == "__main__":
    cumprimentar("Joao")
    valor = calcular_total(12.50, 4)
    print("Total:", valor)
````

> **Por agora, podes usar ou omitir.** Nesta fase ainda não criamos funções nem importamos módulos, por isso podes escrever os teus programas sem este bloco. A partir do manual sobre funções, a convenção começará a fazer todo o sentido.

## 7. PEP 8: guia de estilo para Python
PEP é a sigla de *Python Enhancement Proposal*. A **PEP 8** (Guido van Rossum, 2001) define o guia de estilo que se tornou o padrão da comunidade.

### 7.1. Regras essenciais
- **Indentação:** 4 espaços por nível. Nunca tabulações; nunca misturar.
- **Comprimento de linhas:** máximo 79 caracteres (72 para comentários/docstrings).
- **Linhas em branco:** duas entre definições de funções de topo; uma entre métodos.
- **Espaços em torno de operadores binários** (`=`, `+`, `-`, `==`, ...).
- **Nomes:** `snake_case` para variáveis/funções, `MAIUSCULAS_COM_SUBLINHADO` para constantes, `PascalCase` para classes.

Espaçamento, **bom** versus **mau**:

````python
# Bom:
a, b = 2, 2
x = 5 + 3
if a == b:
    print(a)

# Mau (a evitar):  x=5+3   /   if a==b:
````

### 7.2. Verificadores automáticos
Existem ferramentas que verificam automaticamente a conformidade com a PEP 8: **flake8**, **pycodestyle** e **ruff**. No VS Code, com a extensão Python, podes ativá-las e o editor sublinha os problemas em tempo real.

> **Não te tornes refém da PEP 8.** A própria PEP 8 admite que *"a coerência local é mais importante do que seguir a PEP 8"*. Em código teu, ou na ausência de outra orientação, segue PEP 8.

## 8. Resumo do tópico
- O Python lê o código de cima para baixo, uma instrução por linha.
- A indentação **é sintaxe**: define os blocos de código.
- A norma é **4 espaços** por nível. Nunca misturar Tab e espaços.
- Comentários começam com `#` e explicam o **porquê**, não o **quê**.
- Não há ponto e vírgula no fim das instruções.
- `if __name__ == "__main__":` distingue execução direta de importação.
- A **PEP 8** é o guia de estilo oficial.
